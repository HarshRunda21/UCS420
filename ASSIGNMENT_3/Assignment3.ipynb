```python
employees_data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19",
                     "2013-06-01", "2021-02-10",
                     "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

employees_df = pd.DataFrame(employees_data)

employees_df
```

```python
employees_df.to_csv("employees.csv", index=False)

print("employees.csv created successfully")
```

```python
average_salary = employees_df["Salary"].mean()

print("Average Salary:", average_salary)
```

```python
total_bonus = employees_df["Bonus"].sum()

print("Total Bonus:", total_bonus)
```

```python
youngest_age = employees_df["Age"].min()

print("Youngest Employee's Age:", youngest_age)
```

```python
highest_rating = employees_df["Rating"].max()

print("Highest Performance Rating:", highest_rating)
```

```python
sorted_df = employees_df.sort_values(
    by="Salary",
    ascending=False
)

sorted_df
```

```python
def categorize_rating(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"


employees_df["Performance_Category"] = (
    employees_df["Rating"].apply(categorize_rating)
)

employees_df
```

```python
employees_df.isnull().sum()
```

```python
print("Total missing values:",
      employees_df.isnull().sum().sum())
```

```python
employees_df = employees_df.rename(
    columns={"Employee_ID": "ID"}
)

employees_df
```

```python
more_experience = employees_df[
    employees_df["Years_of_Experience"] > 5
]

more_experience
```

```python
it_employees = employees_df[
    employees_df["Department"] == "IT"
]

it_employees
```

```python
employees_df["Tax"] = employees_df["Salary"] * 0.10

employees_df
```
