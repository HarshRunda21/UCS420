## Q1. Lists

Take the roll number, extract its individual digits, multiply each digit by 10, and perform the required list operations.

```python
# Q1: Create L from the roll number
roll_no = "1024170151"

digits = [int(digit) for digit in roll_no]
L = [digit * 10 for digit in digits]

print("Digits:", digits)
print("L:", L)

# ii. Add two numbers using append() and insert()
L.append(250)
print("After append(250):", L)  # append() adds 250 at the end of the list.

L.insert(2, 350)
print("After insert(2, 350):", L)  # insert() adds 350 at index 2 and shifts later elements right.

# iii. Remove two elements using remove() and pop()
L.remove(250)
print("After remove(250):", L)  # remove() deletes the first occurrence of the value 250.

removed_value = L.pop(2)
print("After pop(2):", L)  # pop() deletes and returns the element at index 2.
print("Popped value:", removed_value)

# iv. Sort ascending and descending
L.sort()
print("Ascending:", L)

L.sort(reverse=True)
print("Descending:", L)

# v. Slicing
print("First three elements:", L[:3])
print("Last three elements:", L[-3:])

# vi. List comprehension: elements greater than the average
average = sum(L) / len(L)
greater_than_average = [x for x in L if x > average]

print("Average:", average)
print("Elements greater than average:", greater_than_average)
```

## Q2. Tuples

Create `scores` using the first 8 values from `L`. Since the roll number has 10 digits, this uses the first 8 values produced in Q1.

```python
# Q2: Create tuple from first 8 values of L
scores = tuple(L[:8])
print("scores:", scores)

# i. Highest score and its index; lowest score and its frequency
highest = max(scores)
highest_index = scores.index(highest)

lowest = min(scores)
lowest_count = scores.count(lowest)

print("Highest score:", highest)
print("Index of highest score:", highest_index)
print("Lowest score:", lowest)
print("Number of times lowest score appears:", lowest_count)

# ii. Reverse the tuple and return it as a list
# Tuples are immutable, so they cannot be reversed in place; a new reversed list is created.
reversed_scores = list(reversed(scores))
print("Reversed tuple as a list:", reversed_scores)

# iii. Ask the user for a score and find its first occurrence
user_score = int(input("Enter a score to search in scores: "))

if user_score in scores:
    print("First occurrence index:", scores.index(user_score))
else:
    print("Score is not present in the tuple.")

# iv. Attempt to change a tuple element
print("\nAttempting scores[0] = 100:")
try:
    scores[0] = 100
except TypeError as error:
    print("Error:", error)
    # Tuples are immutable, so their elements cannot be changed directly; lists are mutable.

# v. Unpacking using * in a single line
first_score, second_score, *remaining_scores = scores
print("First score:", first_score)
print("Second score:", second_score)
print("Remaining scores:", remaining_scores)
```

## Q3. Random Numbers

The roll number is used as the random seed so that the generated list is reproducible and different roll numbers produce different sequences.

```python
import random

# Set the random seed using the roll number
random.seed(1024170151)

# Generate 100 random numbers between 100 and 900 inclusive
random_numbers = [random.randint(100, 900) for _ in range(100)]

print("Random numbers:")
print(random_numbers)

# ii. Count and print all odd numbers
odd_numbers = [x for x in random_numbers if x % 2 != 0]
print("\nOdd numbers:")
print(odd_numbers)
print("Count of odd numbers:", len(odd_numbers))

# iii. Count and print all even numbers
even_numbers = [x for x in random_numbers if x % 2 == 0]
print("\nEven numbers:")
print(even_numbers)
print("Count of even numbers:", len(even_numbers))

# iv. Count and print all prime numbers
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

prime_numbers = [x for x in random_numbers if is_prime(x)]

print("\nPrime numbers:")
print(prime_numbers)
print("Count of prime numbers:", len(prime_numbers))

# v. Number occurring most frequently and its frequency
frequency = {}

for number in random_numbers:
    frequency[number] = frequency.get(number, 0) + 1

most_frequent_number = max(frequency, key=frequency.get)
most_frequent_count = frequency[most_frequent_number]

print("\nMost frequent number:", most_frequent_number)
print("Number of occurrences:", most_frequent_count)
```

## Q4. Sets

Use the same 8 roll-number digits from Q1 **before multiplying by 10**.

```python
# Q4: Create sets A and B from the first 8 digits of the roll number
first_8_digits = digits[:8]

A = {digit * 7 for digit in first_8_digits}
B = {digit * 9 for digit in first_8_digits}

print("First 8 digits:", first_8_digits)
print("Set A:", A)
print("Set B:", B)

# vi. Union
union_set = A.union(B)
print("\nUnion of A and B:", union_set)

# vii. Intersection
intersection_set = A.intersection(B)
print("Intersection of A and B:", intersection_set)

# viii. A - B and B - A using difference()
A_minus_B = A.difference(B)
B_minus_A = B.difference(A)

print("A - B:", A_minus_B)
print("B - A:", B_minus_A)
# difference() gives values present in one set but not the other in the specified direction;
# symmetric_difference() gives values present in either set but not in both.

# ix. Symmetric difference
symmetric_diff = A.symmetric_difference(B)
print("Symmetric difference:", symmetric_diff)

# x. Subset and superset checks
print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))

# xi. Remove X safely using discard()
X = int(input("\nEnter a value X to remove from set A: "))

A.discard(X)
print("Set A after discard:", A)
# discard() is safer than remove() because it does not raise an error if X is absent.
```

## Q5. Dictionaries

Create and modify the dictionary using the specified dictionary operations.

```python
# Q5: Create the original dictionary
my_dict = {
    "name": "Harsh",
    "roll_no": "1024170151",
    "branch": "COPC",
    "age": 21,
    "city": "Patiala"
}

print("Original dictionary:", my_dict)

# i. Rename "city" to "location" without recreating the dictionary
my_dict["location"] = my_dict.pop("city")
print("\nAfter renaming city to location:", my_dict)

# ii. Add CGPA
my_dict["cgpa"] = 7.76
print("After adding CGPA:", my_dict)

# iii. Increase age by 1
my_dict["age"] += 1
print("After increasing age by 1:", my_dict)

# iv. Delete "branch" using pop() and del in two separate copies
dict_pop = my_dict.copy()
removed_branch = dict_pop.pop("branch")

dict_del = my_dict.copy()
del dict_del["branch"]

print("\nCopy after pop():", dict_pop)
print("Value returned by pop():", removed_branch)
print("Copy after del:", dict_del)
# pop() removes a key and returns its value, while del removes it without returning the value.

# v. Iterate using .items()
print("\nKey-value pairs:")
for key, value in my_dict.items():
    print(f"{key} → {value}")

# vi. Check whether "email" exists before accessing it
if "email" in my_dict:
    print("\nEmail:", my_dict["email"])
else:
    print("\nEmail is not present in the dictionary.")

# vii. Create a friend's dictionary and merge using {**dict1, **dict2}
friend_dict = {
    "name": "Rahul",
    "roll_no": "1024170160",
    "branch": "CSE",
    "age": 21,
    "city": "Chandigarh"
}

merged_dict = {**my_dict, **friend_dict}

print("\nFriend dictionary:", friend_dict)
print("Merged dictionary:", merged_dict)
# When both dictionaries have the same key, values from the second dictionary win.

# viii. Dictionary comprehension: only key-value pairs whose values are strings
string_values_dict = {
    key: value for key, value in my_dict.items()
    if isinstance(value, str)
}

print("\nDictionary containing only string values:", string_values_dict)
```
