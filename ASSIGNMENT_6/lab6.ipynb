```python
## Q1. Vectorized operations and Boolean indexing

import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

# a) Add 2°C to every reading using vectorization
adjusted = temperature + 2
print("a) After adding 2°C:", adjusted)

# b) Convert Celsius to Fahrenheit
fahrenheit = (9 / 5) * temperature + 32
print("b) Temperature in Fahrenheit:", fahrenheit)

# c) Identify readings greater than 32°C using Boolean indexing
above_32 = temperature > 32
print("c) Readings > 32°C:", temperature[above_32])

# d) Count how many readings exceed 32°C
count_above_32 = np.count_nonzero(temperature > 32)
print("d) Count of readings > 32°C:", count_above_32)

# e) Explanation
print("\ne) Vectorization means operations are applied to the entire array at once in optimized NumPy code.")
print("   Boolean indexing creates a mask of True/False values to select only the matching elements.")
print("   This avoids Python-level loops and is much faster for large datasets.")
```

```python
## Q2. Matrix operations on daily steps

import numpy as np

steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print("a) Total steps recorded:", steps.sum())
print("b) Mean number of steps:", steps.mean())
print("c) Maximum value:", steps.max())
print("   Minimum value:", steps.min())
print("d) Total steps for each day (axis=0):\n", steps.sum(axis=0))
print("e) Total steps for each user (axis=1):\n", steps.sum(axis=1))
print("f) Position of maximum number of steps:", np.unravel_index(np.argmax(steps), steps.shape))
```

```python
## Q3. Slicing, copying, and reshaping with NumPy

import numpy as np

original = np.array([1, 2, 3, 4, 5, 6])
subset = original[1:5]
subset[0] = 999
print("c) After modifying subset:")
print("original:", original)
print("subset:", subset)

copied = original[1:5].copy()
copied[0] = 500
print("\nd) After modifying .copy() slice:")
print("original:", original)
print("copied:", copied)

matrix = np.arange(1, 13).reshape(3, 4)
print("\ne) 3x4 matrix:\n", matrix)
print("First row:", matrix[0])
print("Last row:", matrix[-1])
print("Second column:", matrix[:, 1])
print("Rows 1-2 and columns 2-3:\n", matrix[1:3, 1:3])

flat_ravel = matrix.ravel()
flat_flatten = matrix.flatten()
print("\ng) Flattened using ravel():", flat_ravel)
print("Flattened using flatten():", flat_flatten)

flat_ravel[0] = 1000
print("\nh) After changing ravel() output:")
print("original matrix:\n", matrix)
print("ravel view:", flat_ravel)

flat_flatten[0] = 2000
print("\ni) After changing flatten() output:")
print("original matrix:\n", matrix)
print("flatten copy:", flat_flatten)

print("\nj) Matrix properties:")
print("shape:", matrix.shape)
print("ndim:", matrix.ndim)
print("size:", matrix.size)
print("dtype:", matrix.dtype)
```

```python
## Q4. OLS for a cognitive-assistive prediction model

import numpy as np

X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])
y = np.array([40, 65, 30, 85])

print("a) Shape of X:", X.shape)
print("   Dimensions of X:", X.ndim)

X_transpose = X.T
print("\nb) X.T:\n", X_transpose)
print("   The transpose swaps rows and columns; here each original column becomes a row, which is useful in matrix algebra and linear regression.")

XtX = X.T @ X
print("\nc) X.T @ X:\n", XtX)

# d) Matrix inverse where appropriate
print("\nd) Inverse of X.T @ X:")
print(np.linalg.inv(XtX))

# e) OLS coefficients beta = (X^T X)^-1 X^T y
beta = np.linalg.inv(X.T @ X) @ X.T @ y
print("\ne) OLS coefficients beta:", beta)

# f) Interpretation
print("\nf) Coefficients represent how each feature affects the prediction output, while keeping the others fixed.")
print("   In this context, they show the weight of sleep hours, activity level, and stress level in estimating the assistance score.")

new_user = np.array([5, 40, 7])
prediction = new_user @ beta
print("\ng) Predicted assistance score for new_user:", prediction)
```
