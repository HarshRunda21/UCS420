This notebook contains the solutions for all five NumPy questions.


```python
## Q1. Basic NumPy operations on a 1D array

import numpy as np

arr = np.array([1, 2, 3, 4, 5])
print("Original array:", arr)
print("After +2:", arr + 2)
print("After *3:", arr * 3)
print("After /2:", arr / 2)
```

```python
## Q2. Reverse array and find most frequent values

arr = np.array([1, 2, 3, 6, 4, 5])
print("Reversed array:", arr[::-1])

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

for name, arr2 in [("x", x), ("y", y)]:
    values, counts = np.unique(arr2, return_counts=True)
    max_count = counts.max()
    frequent = values[counts == max_count]
    indices = []
    for value in frequent:
        indices.extend(np.where(arr2 == value)[0].tolist())
    print(f"\nArray {name}:")
    print("Most frequent value(s):", frequent)
    print("Indices:", indices)
```

```python
## Q3. Access array elements by row and column

arr = np.array([[10, 20, 30],
                [40, 50, 60],
                [70, 80, 90]])

print("1st row, 2nd column:", arr[0, 1])
print("3rd row, 1st column:", arr[2, 0])
```

```python
## Q4. Create an evenly spaced 1D array and inspect its properties

import numpy as np

harsh_arr = np.linspace(10, 100, 25)
print("Array:", harsh_arr)
print("Dimensions:", harsh_arr.ndim)
print("Shape:", harsh_arr.shape)
print("Total elements:", harsh_arr.size)
print("Data type of each element:", harsh_arr.dtype)
print("Bytes consumed by the array:", harsh_arr.nbytes)

arr_2d = harsh_arr.reshape(5, 5)
print("Transpose using reshape():")
print(arr_2d.T)
print("T attribute on a 1D array:", harsh_arr.T)
print("For 1D arrays, .T has no real transpose effect; it remains the same array.")
```

```python
## Q5. Create, reshape, and resize a 2D NumPy array

import numpy as np

ucs420_harsh = np.array([[10, 20, 30, 40],
                         [50, 60, 70, 80],
                         [90, 15, 20, 35]])

print("Original array:\n", ucs420_harsh)
print("Mean:", np.mean(ucs420_harsh))
print("Median:", np.median(ucs420_harsh))
print("Max:", np.max(ucs420_harsh))
print("Min:", np.min(ucs420_harsh))
print("Unique elements:", np.unique(ucs420_harsh))

reshaped_ucs420_harsh = ucs420_harsh.reshape(4, 3)
print("\nReshaped to 4x3:\n", reshaped_ucs420_harsh)

resized_ucs420_harsh = np.resize(ucs420_harsh, (2, 3))
print("\nResized to 2x3:\n", resized_ucs420_harsh)
```
